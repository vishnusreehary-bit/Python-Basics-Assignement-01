```python
# 1. Simple ATM Calculator
balance = float(input("Enter account balance: "))
withdrawal = float(input("Enter withdrawal amount: "))

if withdrawal <= balance:
    balance = balance - withdrawal
    print("Transaction successful. Remaining balance: ₹", balance)
else:
    print("Insufficient balance.")
```

```python
# 2. Electricity Bill Generator
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5

elif units <= 200:
    bill = (100 * 5) + ((units - 100) * 7)

else:
    bill = (100 * 5) + (100 * 7) + ((units - 200) * 10)

print("Bill amount = ₹", bill)
```

```python
# 3. Temperature Converter
temp = float(input("Enter temperature: "))
unit = input("Enter unit (C/F): ")

if unit.upper() == "C":
    fahrenheit = (temp * 9/5) + 32
    print("Temperature in Fahrenheit:", fahrenheit, "°F")

elif unit.upper() == "F":
    celsius = (temp - 32) * 5/9
    print("Temperature in Celsius:", celsius, "°C")

else:
    print("Invalid unit.")
```

```python
# 4. Mobile Recharge Validity
from datetime import datetime

recharge_date = input("Enter recharge date (YYYY-MM-DD): ")
today = input("Enter today's date (YYYY-MM-DD): ")

recharge_date = datetime.strptime(recharge_date, "%Y-%m-%d")
today = datetime.strptime(today, "%Y-%m-%d")

days = (today - recharge_date).days

if days < 28:
    print("Plan is still valid.")
else:
    print("Plan expired. Recharge again.")
```

```python
# 5. Bus Seat Reservation System
booked = ["Amit", "Sara", "John"]

cancel = input("Enter passenger name to cancel: ")
new_passenger = input("Enter new passenger name: ")
position = int(input("Enter position: "))

booked.remove(cancel)
booked.insert(position, new_passenger)

print("Updated booking list:", booked)
```

```python
# 6. Sum of Even Numbers
number = int(input("Enter a number: "))

total = 0

for i in range(1, number + 1):
    if i % 2 == 0:
        total = total + i

print("Sum of even numbers:", total)
```

```python
# 7. Dictionary – Student Marks Summary
students = {}

for i in range(3):
    name = input("Enter name: ")
    marks = float(input("Enter marks: "))

    students[name] = marks

average = sum(students.values()) / len(students)

print("Student Marks:", students)
print("Average Marks:", average)
```

```python
# Dictionary - Word Frequency
sentence = input("Enter a sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

for word, count in frequency.items():
    print(word, ":", count)
```

```python
# 9. Function – Check Prime
def check_prime(number):

    if number <= 1:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True


number = int(input("Enter a number: "))

if check_prime(number):
    print(number, "is a prime number.")
else:
    print(number, "is not a prime number.")
```

```python
# 10. Vowel Counter
def count_vowels(text):

    count = 0

    for character in text:
        if character.lower() in "aeiou":
            count = count + 1

    return count


text = input("Enter a string: ")

result = count_vowels(text)

print("Number of vowels:", result)
```

```python
# 11. Count Divisible Numbers
N = int(input("Enter the range limit (N): "))
D = int(input("Enter the divisor (D): "))

count = 0

for i in range(1, N + 1):
    if i % D == 0:
        count = count + 1

print("Count of numbers divisible by", D, ":", count)
```
